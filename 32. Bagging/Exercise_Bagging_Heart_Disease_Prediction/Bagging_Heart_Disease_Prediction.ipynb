#### Handle text columns using label encoding and one hot encoding

```python
df4 = df3.copy()

df4 = df4.replace({
    'Exercise_Angina': {'N':0, 'Y':1},
    'ST_Slope': {'Down':1, 'Flat':2,'Up':3},
    'RestingECG': {'Normal':1, 'ST':2, 'LVH':3}
})

df4 = df4.infer_objects(copy=False)
```

```python
df4.head()
```

```python
df5 = pd.get_dummies(df4, drop_first=True, dtype=int)
df5.head()
```

When Should You Use drop_first=True?

1. ✔ Yes → When using Linear / Logistic Regression
1. ❌ No → When using Tree models (Random Forest, XGBoost)

Interview-Level Explanation

If asked:
Why drop_first=True?

You answer:

To prevent multicollinearity caused by redundant dummy variables, especially when using linear models.

That’s the correct reasoning.

Train Test Split

```python
X = df5.drop("HeartDisease", axis='columns')
y = df5.HeartDisease

X.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
X_scaled[:3]
```

```python
# from sklearn.model_selection import train_test_split

# X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, stratify=y, random_state=10)
```

Training a model using SVM later bagging 

```python
from sklearn.svm import SVC
from sklearn.model_selection import cross_val_score

scores = cross_val_score(SVC(), X, y, cv=5)
scores.mean()
```

**Use bagging now with SVM**

```python
from sklearn.ensemble import BaggingClassifier

bag_model = BaggingClassifier(estimator=SVC(), n_estimators=100, max_samples=0.8, random_state=20)
scores = cross_val_score(bag_model, X, y, cv=5)
scores.mean()
```

As you can see above, using bagging in case of SVM doesn't make much difference in terms of model accuracy. Bagging is effective when we have high variance and instable model such as decision tree. Let's explore how bagging changes the performance for a decision tree classifier.

#### Train a model using decision tree and then using bagging

```python
from sklearn.tree import DecisionTreeClassifier

scores = cross_val_score(DecisionTreeClassifier(random_state=0), X, y, cv=5)
scores.mean()
```

```python
bag_model = BaggingClassifier(
                              estimator=DecisionTreeClassifier(random_state=0), 
                              n_estimators=100,
                              max_samples=0.9,
                              oob_score=True,
                              random_state=20
                              )

scores = cross_val_score(bag_model, X, y)
scores
```

```python
scores.mean()
```

**with bagging the score improved from 71.93% to 79.81%**

#### Train a model using Random Forest which itself uses bagging underneath

```python
from sklearn.ensemble import RandomForestClassifier

scores = cross_val_score(RandomForestClassifier(), X, y, cv=5)
scores.mean()
```

**Random forest gave even a better performance with 82.25% as score. Underneath it used bagging where it sampled not only data rows but also the columns (or features)**

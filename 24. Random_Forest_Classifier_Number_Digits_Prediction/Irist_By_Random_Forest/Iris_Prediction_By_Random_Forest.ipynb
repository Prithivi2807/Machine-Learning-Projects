Use famous iris flower dataset from sklearn.datasets to predict flower species using random forest classifier.
1. Measure prediction score using default n_estimators (10)
2. Now fine tune your model by changing number of trees in your classifer and tell me what best score you can get using how many trees

```python
from sklearn.datasets import load_iris

# Load the dataset 
dataset = load_iris(as_frame = True)
print(dataset)
```

```python
dir(dataset)
```

```python
X = dataset.data
y = dataset.target
print(f"Input shape: {X.shape}")
print(f"Outpus shape: {y.shape}")
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
from sklearn.ensemble import RandomForestClassifier
model = RandomForestClassifier(n_estimators=100)
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

according to iris data set random force classifier i applied n estimator by default 10 means it's coming nine percent accuracy but i havebeen changed n estimator=100 so if get 100 it gives 96 percentage of the output score uh this is pretty much good accuracy to predict the flowers.

```python
y_predicted = model.predict(X_test)
```

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_predicted)
cm
```

```python
%matplotlib inline
import matplotlib.pyplot as plt
import seaborn as sn 
plt.figure(figsize=(10, 7))
sn.heatmap(cm, annot=True)
plt.xlabel('Predicted')
plt.ylabel('Truth')
```

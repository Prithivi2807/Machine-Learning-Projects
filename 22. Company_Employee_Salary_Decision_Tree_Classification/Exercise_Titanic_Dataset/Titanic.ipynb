```python
import pandas as pd
```

```python
df = pd.read_csv(r"E:\1_Data_Science\Machine-Learning-Projects\22. Company_Employee_Salary_Decision_Tree_Classification\Exercise_Titanic_Dataset\titanic.csv")
df
```

```python
df.columns
```

```python
df.isna().sum()
```

```python
df['Age'] = df['Age'].fillna(df['Age'].mean())
```

```python
df.isna().sum()
```

```python
X = df[['Pclass', 'Sex', 'Age', 'Fare']].copy()
X
```

```python
X['Age'] = X['Age'].round().astype(int)
```

```python
X
```

```python
# from sklearn.preprocessing import LabelEncoder
# le_sex = LabelEncoder()
```

```python
# X['sex_n'] = le_sex.fit_transform(X['Sex'])
```

```python
X.Sex = X.Sex.map({'male':1, 'female':2})
```

```python
X
```

```python
# X = X.drop(['Sex'], axis='columns')
```

```python
X
```

```python
y = df['Survived']
y
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, train_size=0.8)
```

```python
len(X_train)
```

```python
len(X_test)
```

```python
from sklearn import tree
model= tree.DecisionTreeClassifier()
```

```python
model.fit(X_train, y_train)
```

```python
model.score(X_test, y_test)
```

```python
y_preds = model.predict(X_test)
```

In titanic Dataset our model score is 80 percent

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_preds)
cm
```

```python
import seaborn as sn 
import matplotlib.pyplot as plt

plt.figure(figsize=(10, 7))
sn.heatmap(cm, annot=True, fmt='d',
           xticklabels=['Not Survived', 'Survived'],
           yticklabels=['Not Survided', 'Survived'])

plt.xlabel('Predicted')
plt.ylabel('Actual')
```

```python
# Where:

# Top-left (TN) → correctly predicted not survived

# Top-right (FP) → predicted survived but didn’t

# Bottom-left (FN) → predicted not survived but did

# Bottom-right (TP) → correctly predicted survived
```
